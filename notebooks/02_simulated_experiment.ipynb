# Independent simulated A/B experiment
No GA4 results. Seed 42; one assignment and binary outcome per synthetic user. Two-sided alpha .05; no peeking or multiple primary outcomes.

```python
from productpulse.experiment import simulate, analyze
experiment = simulate(seed=42)
experiment.groupby('treatment').converted.agg(['count', 'sum', 'mean'])
```

```python
analyze(experiment)
```

```python
experiment.groupby(['treatment', 'mobile']).converted.mean().unstack().plot.bar(title='SIMULATED conversion by assignment/device')
```
